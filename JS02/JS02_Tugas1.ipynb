PISAHKAN VARIABEL YANG DAPAT DAN TIDAK DAPAT DI PISAHKAN

```python
# Variabel yang tidak digunakan
unused_columns = ['id', 'Unnamed: 32']

# Hapus variabel yang tidak digunakan
df = df.drop(columns=unused_columns)

df.head()
```

```python
print("Kolom yang digunakan:")
print(df.columns.tolist())
```

ENCODING KOLOM DIAGNOSIS



```python
df['diagnosis'] = df['diagnosis'].map({
    'M': 1,
    'B': 0
})

df.head()
```

```python
df['diagnosis'].value_counts()
```

PISAHKAN FITUR DAN TARGET

```python
X = df.drop(columns=['diagnosis'])
y = df['diagnosis']
```

STANDARISASI SEMUA YANG NUMERIK



```python
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

```python
X_scaled = pd.DataFrame(
    X_scaled,
    columns=X.columns
)

X_scaled.head()
```

GABUNGKAN DENGAN DIAGNOSIS

```python
df_scaled = X_scaled.copy()
df_scaled['diagnosis'] = y.values

df_scaled.head()
```
