## Langkah 2: Import Library

NumPy dan Pandas digunakan untuk pengolahan dan manipulasi data.


```python
import numpy as np
import pandas as pd
```

## Langkah 3: Baca Data

Membaca data dari file CSV menggunakan Pandas.


```python
data = pd.read_csv('dataset.csv')
data.head()
```

## Langkah 4: Pemahaman Terhadap Data

Menampilkan data awal, ukuran data, informasi kolom, dan statistik deskriptif untuk memahami karakteristik dataset.


```python
# Melihat beberapa data awal
display(data.head())

# Mengecek ukuran data
print("Ukuran data:", data.shape)

# Informasi tentang data
print("\nInformasi data:")
data.info()

# Deskripsi statistik data
display(data.describe())
```

## Langkah 5: Visualisasi Data

Matplotlib dan Seaborn digunakan untuk melihat hubungan antarvariabel dan korelasi pada data.


```python
import matplotlib.pyplot as plt
import seaborn as sns
```

### Pairplot

Pairplot digunakan untuk menampilkan hubungan antara beberapa variabel bebas dengan `Yearly Amount Spent`.


```python
sns.pairplot(
    data,
    x_vars=['Time on App', 'Time on Website', 'Length of Membership'],
    y_vars='Yearly Amount Spent',
    height=4,
    aspect=1,
    kind='scatter'
)
plt.show()
```

### Heatmap Korelasi

```python
numeric_data = data.select_dtypes(include=np.number)

plt.figure(figsize=(9, 6))
sns.heatmap(numeric_data.corr(), cmap="YlGnBu", annot=True)
plt.title("Matriks Korelasi")
plt.show()
```

## Langkah 6: Regresi Linier

Variabel `Length of Membership` digunakan sebagai variabel bebas (**X**) dan `Yearly Amount Spent` sebagai variabel target (**y**).


```python
X = data['Length of Membership']
y = data['Yearly Amount Spent']

print("X:")
display(X.head())

print("y:")
display(y.head())
```

### Membagi Data Latih dan Data Uji


```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    train_size=0.7,
    test_size=0.3,
    random_state=100
)

print("Jumlah data latih:", len(X_train))
print("Jumlah data uji:", len(X_test))
```

### Training Model

Model regresi linier dilatih menggunakan **Ordinary Least Squares (OLS)** dari StatsModels.
Konstanta ditambahkan agar model memiliki intercept.


```python
import statsmodels.api as sm

X_train_sm = sm.add_constant(X_train)
lr = sm.OLS(y_train, X_train_sm).fit()

print(lr.summary())
```

### Visualisasi Garis Regresi pada Data Latih

Garis regresi dibuat berdasarkan intercept dan koefisien hasil training model.


```python
intercept = lr.params['const']
slope = lr.params['Length of Membership']

plt.scatter(X_train, y_train)
plt.plot(X_train, intercept + slope * X_train, 'r')
plt.xlabel('Length of Membership')
plt.ylabel('Yearly Amount Spent')
plt.title('Regresi Linier pada Data Latih')
plt.show()

print("Intercept:", intercept)
print("Koefisien:", slope)
```

## Langkah 7: Analisis Residual

Residual adalah selisih antara nilai sebenarnya dan nilai hasil prediksi model:

**Residual = y aktual - y prediksi**


```python
# Prediksi nilai y pada data latih
y_train_pred = lr.predict(X_train_sm)

# Menghitung residual
res = y_train - y_train_pred

display(res.head())
```

### Histogram Residual

Histogram digunakan untuk melihat distribusi error model.


```python
plt.figure(figsize=(7, 5))
sns.histplot(res, bins=15, kde=True)
plt.title('Error Terms', fontsize=15)
plt.xlabel('y_train - y_train_pred', fontsize=15)
plt.show()
```

### Scatter Plot Residual

Scatter plot residual digunakan untuk melihat apakah terdapat pola tertentu pada error.


```python
plt.scatter(X_train, res)
plt.axhline(y=0, linestyle='--')
plt.xlabel('Length of Membership')
plt.ylabel('Residual')
plt.title('Residual Plot')
plt.show()
```

## Langkah 8: Prediksi pada Data Uji dan Evaluasi Model

Model yang telah dilatih digunakan untuk memprediksi nilai target pada data uji.


```python
X_test_sm = sm.add_constant(X_test)
y_test_pred = lr.predict(X_test_sm)

display(pd.DataFrame({
    'Aktual': y_test,
    'Prediksi': y_test_pred
}).head())
```

### Menghitung R-squared

Nilai **R² (R-squared)** digunakan untuk mengukur seberapa besar variasi pada target yang dapat dijelaskan oleh model.


```python
from sklearn.metrics import r2_score

r_squared = r2_score(y_test, y_test_pred)
print("R-squared:", r_squared)
```

## Langkah 9: Visualisasi Hasil

Menampilkan data uji dan garis hasil prediksi model regresi.


```python
# Urutkan X agar garis prediksi terlihat rapi
hasil = pd.DataFrame({
    'X_test': X_test,
    'y_test': y_test,
    'y_test_pred': y_test_pred
}).sort_values('X_test')

plt.scatter(hasil['X_test'], hasil['y_test'])
plt.plot(hasil['X_test'], hasil['y_test_pred'], 'r')
plt.xlabel('Length of Membership')
plt.ylabel('Yearly Amount Spent')
plt.title('Hasil Prediksi pada Data Uji')
plt.show()
```
